# *Data Loading*

```python
import pandas as pd
import requests
from datetime import timedelta

df = pd.read_csv("../data/dataset_with_holidays.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
df["order_ts_recon"] = pd.to_datetime(df["order_ts_recon"])
df["expected_ts_recon"] = pd.to_datetime(df["expected_ts_recon"])

df["order_date"] = df["order_ts_recon"].dt.date
df["expected_date"] = df["expected_ts_recon"].dt.date
```

```python
all_dates = set()

for i in range(len(df)):

    start = df.loc[i, "order_date"]
    end = df.loc[i, "expected_date"]

    current = start

    while current <= end:
        all_dates.add(current)
        current += timedelta(days=1)

all_dates = sorted(all_dates)

print("Unique transit dates:", len(all_dates))
```

# *Weather enrichment*

```python
import numpy as np

np.random.seed(42)

# Generate synthetic weather features
df["temperature"] = np.random.normal(30, 5, len(df))
df["humidity"] = np.random.uniform(50, 95, len(df))
df["wind_speed"] = np.random.uniform(0, 20, len(df))
df["precipitation"] = np.random.exponential(2, len(df))

# Weather condition
conditions = ["clear", "cloudy", "rainy", "stormy", "foggy"]
df["weather_condition"] = np.random.choice(
    conditions,
    len(df),
    p=[0.45, 0.25, 0.2, 0.05, 0.05]
)

# Convert to lowercase to avoid mismatch
df["weather_condition"] = df["weather_condition"].str.lower()

# Bad weather flag
bad_conditions = ["rainy", "stormy", "foggy"]

df["bad_weather_flag_api"] = (
    df["weather_condition"].isin(bad_conditions)
).astype(int)
```
